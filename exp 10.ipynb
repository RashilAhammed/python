```python
#count the occurrances of each word ina line of text
text = input("enter a line of text : ")
words = text.split()
count = {}
for word in words:
    count[word] = count.get(word, 0) + 1
print("word occurrences:", count)
```

```python
#count the occurance of a in a list of first names
names = input("Enter first names separated by space: ").split()
count = 0
for name in names:
    count = count + name.lower().count("a")
print("Number of 'a' : ",count)
```

```python
#find gcd of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter first number: "))
while b != 0:
    a, b = b, a % b
print("GCD =" , a)    
```

```python
#create a list by removing even numbers
numbers = list(map(int, input("Enter intergers separated by space:").split()))
result = []
for n in numbers:
    if n % 2 != 0:
        result.append(n)
print("list after renoving even numbers:", result)        
```

```python
#find the factorial of a number
n = int(input("Enter a number: "))
factorial = 1
for i in range(1, n + 1):
    factorial = factorial * i
print("Factorial =", factorial)    
```

```python
#Four-digit numbers with all digits even and perfect square
start = int(input("Enter starting value: "))
end = int(input("Enter ending value: "))

import math

result = []

for n in range(start, end + 1):
    if 1000 <= n <= 9999:
        digits = str(n)

        if all(int(d) % 2 == 0 for d in digits):
            if math.isqrt(n) ** 2 == n:
                result.append(n)

print("Result:", result)
```

```python
#Count character frequency in a string
text = input("Enter a string: ")

frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

print("Character frequency:", frequency)
```

```python
#Compare two lists
list1 = list(map(int, input("Enter first list: ").split()))
list2 = list(map(int, input("Enter second list: ").split()))

if len(list1) == len(list2):
    print("Lists have same length")
else:
    print("Lists do not have same length")

if sum(list1) == sum(list2):
    print("Lists have same sum")
else:
    print("Lists do not have same sum")

common = []

for n in list1:
    if n in list2:
        common.append(n)

if common:
    print("Common values:", common)
else:
    print("No common values")
```

```python
#Pyramid pattern
n = int(input("Enter N: "))

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i * j, end=" ")
    print()
```
