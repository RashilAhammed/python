```python
r1=int(input("enter roll number 1: "))
n1=input("enter name 1:")
r2=int(input("enter roll number 2: "))
n2=input("enter name 2: ")
r3=int(input("enter roll number 3: "))
n3=input("enter name 3: ")
r4=int(input("enter roll number 4: "))
n4=input("enter name 4: ")
r5=int(input("enter roll number 5: "))
n5=input("enter name 5: ")
students={r1:n1,r2:n2,r3:n3,r4:n4,r5:n5}
print("dictionary=",students)
ascending=dict(sorted(students.items()))
descending=dict(sorted(students.items(),reverse=True))
print("ascending order=",ascending)
print("descenting order=",descending)
```
