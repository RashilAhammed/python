```python
#change a given string to a new string where the first and last chars have been changed
str=input("enter a string")
new_str= str[-1]+str[1:-1]+str[0]
print("new string:",new_str)
```

```python
#string made of the first 2 and last 2 chars 
str=input("enter a string")
result=str[:2]+str[-2:]
print("new string",result)
```

```python
#replace occurance of first charecter to $
str1 = input("Enter a string: ")

first = str1[0]
result = first + str1[1:].replace(first, "$")

print("Result:", result)
```
