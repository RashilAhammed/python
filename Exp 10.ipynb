```python
#count the occurance
text=input("enter a line of text")
words=text.split()
count={}
for word in words:
     count[word]= count.get(word,0)+1
print("word occurances:")
print(count)
```

```python
#replace integers greater than 100 with "over" ina list
numbers=input("enter numbers seperated by commas:").split(",")
result=[]
for n in numbers:
    n=int(n)
    
    if n>100:
        result.append("over")
    else:
        result.append(n)

print(result)
```

```python
#count the occurance of a within the list
names=input("eneter list first names :").split(",")
count=0
for name in names:
    count= count+name.lower().count("a")
print("number of a count",count)
```

```python
#GCD of 2 numbers
a=int(input("enter first number:"))
b=int(input("enter second number:"))
while b!=0:
    a,b=b,a%b
print("GCD=",a)
```
