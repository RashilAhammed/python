```python
#comma-separated numbers from user and generate list and find sum
numbers=input("enter comma separated number")
numberlist=numbers.split(",")
sum=0
for i in numberlist:
    sum=sum+int(i)
print("list:",numberlist)
print("sum:",sum)
```
