```python
#get the 3rd element and 3rd element from last of the tuple
t=tuple(input("enter elements seperated by comma: ").split(","))
print("3rd element :",t[2])
print("3rd element from last:",t[-3])
```

```python
#generate a tuple from comma seperated numbers and find its average
numbers=input("enter number seperated by commmas: ")
numbers=numbers.split(",")
a=int(numbers[0])
b=int(numbers[1])
c=int(numbers[2])
numbers=(a,b,c)
average=sum(numbers)/len(numbers)
print("tuple=",numbers)
print("average=",average)
```
