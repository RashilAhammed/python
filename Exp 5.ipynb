```python
#get the 3rd element and 3rd element from last of the tuple
tuple=(1,2,3,4,5,6)
print("3rd element :",tuple[2])
print("3rd element from last:",tuple[-3])
```

```python
#generate a tuple from comma seperated numbers and find its average
number=input("enter comma seperated numbers:")
t=tuple(map(int,number.split(",")))
average=sum(t)/len(t)
print("tuple:",t)
print("average:",average)
```
